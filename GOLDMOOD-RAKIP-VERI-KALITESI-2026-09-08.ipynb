# GoldMood rakip veri kalitesi
Birim: tenant içindeki rakip kaydı; Instagram ölçüsü 12 gönderilik örneklem.
Bing 10 doğum sorgusundaki ilgisizlik başlık/alan adı incelemesiyle saptandı; aşağıdaki sayı yalnız sorgu öneki sayımıdır, otomatik semantik kalite puanı değildir.
Kaynak JSON kişisel müşteri bilgisi içermez. Para birimi ve Stripe sipariş mutabakatı bu notebook kapsamı dışındadır.

```python
import json
from pathlib import Path
p = Path("GOLDMOOD-RAKIP-STRATEJI-KANITLARI-2026-09-08.json")
d = json.loads(p.read_text())
rows=d["reports"]
assert len(rows)==30
assert len({x["id"] for x in rows})==30
print("Reviewed",sum(x["status"]=="reviewed" for x in rows))
print("Insufficient",sum(x["status"]=="insufficient" for x in rows))
print("Bing doğum-prefix queries",sum(x["query"].startswith("doğum") for x in d["bingQueries"]))
for x in rows:
    if x["stats"]:
        s=x["stats"]
        print(x["name"],s["known_like_count"],"/",s["sample_posts"],s["first_post"],s["last_post"],s["median_likes_known"])
print("GSC",d["gsc"]["range"],d["gsc"]["totals"])
```
